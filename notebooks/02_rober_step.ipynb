```python
QUICK = True
N_TRAJ, N_PAIRS, EPOCHS = (20, 20, 3) if QUICK else (2000, 50, 150)
N_WORKERS = 1
```

## 1. Generate data

```python
import numpy as np
from chemistry_emulator.systems.rober import make_step_data

path = f"{OUT}/rober_step_{N_TRAJ}x{N_PAIRS}.npz"
if os.path.exists(path):
    d = np.load(path); X, Y, TID = d["X"], d["Y"], d["TID"]
else:
    X, Y, TID = make_step_data(N_TRAJ, N_PAIRS, seed=0, n_workers=N_WORKERS)
    np.savez(path, X=X, Y=Y, TID=TID)
print(X.shape, Y.shape)
```

## 2. Train and compare with the 'copy y(t)' baseline

```python
from chemistry_emulator.training import prepare, train, save_run
from chemistry_emulator.models import MLP, DeepONet
from chemistry_emulator.evaluate import evaluate_step

data = prepare(X, Y, TID, seed=0)
models = {"mlp": MLP(n_hidden=5, n_in=4, n_out=3),
          "deeponet": DeepONet(n_hidden=5, n_out=3, n_branch_in=3)}
params = {}
for name, m in models.items():
    params[name] = train(m, data, epochs=EPOCHS)
    save_run(f"{OUT}/rober_step_{name}.pkl", params[name], data)
    evaluate_step(m, params[name], data)
```

## 3. Rollout versus the solver
Start at the standard ROBER state, step with a fixed dt, and watch the error grow.

```python
from chemistry_emulator.evaluate import rollout_vs_solver
from chemistry_emulator.systems.rober import solve_trajectory, K_REF
from chemistry_emulator.plotting import plot_rollout_error

DT, N_STEPS = 0.1, 100
y0 = [1 - 1e-6, 1e-6, 0.0 + 1e-10]
for name, m in models.items():
    times, pred, true, rel = rollout_vs_solver(m, params[name], data, solve_trajectory, K_REF, y0, DT, N_STEPS)
    print(name, "final rel. error per species:", rel[-1], "| sum(y)-1:", pred[-1].sum() - 1)
    plot_rollout_error(times, rel).axes[0].set_title(name)
```
